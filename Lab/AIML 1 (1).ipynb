Vacuum Cleaner Agent Using Artificial Intelligence(Simple Reflex Agent)

Aim:
To design and implement a simple AI-based laboratory vacuum cleaner agent that can move between different laboratory areas, detect dirty locations, clean them according to their priority, and manage its battery while performing cleaning tasks.



---
Experimental Objective:

*   To represent a laboratory as an environment.
*   To implement a vacuum cleaner as an intelligent agent.
*   To detect dirt and occupancy in different laboratory sections.


---

*   Create an agent that can observe and modify its environment.
*   Implement simple decision-making logic.

Concept:

An AI agent is a system that observes its environment, makes decisions, and performs actions to achieve a particular goal.

In this experiment:


1.   Environment → Laboratory
2.   Agent → Vacuum Cleaner
1.   Goal → Clean dirty laboratory areas
2.   Sensor → Detect dirt level and occupancy
1.   Action → Move and clean
2.   Performance → Cleaning score

```python
import random


# Create Laboratory Environment
class LaboratoryEnvironment:
    def __init__(self):
        self.areas = {
            "Chemistry Laboratory": {
                "dirt": random.randint(20, 90),
                "priority": 5,
                "occupied": random.choice([True, False])
            },

            "Physics Laboratory": {
                "dirt": random.randint(20, 90),
                "priority": 4,
                "occupied": random.choice([True, False])
            },

            "Computer Laboratory": {
                "dirt": random.randint(20, 90),
                "priority": 3,
                "occupied": random.choice([True, False])
            },

            "Research Laboratory": {
                "dirt": random.randint(20, 90),
                "priority": 5,
                "occupied": random.choice([True, False])
            },

            "Laboratory Corridor": {
                "dirt": random.randint(20, 90),
                "priority": 2,
                "occupied": False
            }
        }


# Laboratory Vacuum Cleaner Agent
class LaboratoryVacuumAgent:
    def __init__(self, environment):
        self.environment = environment
        self.currentLocation = "Laboratory Corridor"
        self.battery = 100
        self.score = 0


    # Display laboratory environment before cleaning
    def display_environment(self):
        print("\nLaboratory Environment")
        print("-----------------------------------")

        for area, data in self.environment.areas.items():
            print(
                area,
                "--> Dirt:", data["dirt"],
                ", Priority:", data["priority"],
                ", Occupied:", data["occupied"]
            )


    # Calculate priority score
    def calculate_score(self, area):
        data = self.environment.areas[area]

        if data["occupied"]:
            return -1

        return data["dirt"] + (data["priority"] * 10)


    # Select the best laboratory area to clean
    def select_area(self):
        best_area = None
        best_score = -1

        for area in self.environment.areas:

            if self.environment.areas[area]["dirt"] > 0:

                area_score = self.calculate_score(area)

                if area_score > best_score:
                    best_score = area_score
                    best_area = area

        return best_area


    # Move vacuum cleaner to selected laboratory area
    def move_to(self, area):
        print(
            "\nMoving from",
            self.currentLocation,
            "to",
            area
        )

        self.currentLocation = area

        self.battery -= 10
        self.score -= 1


    # Clean selected laboratory area
    def clean_area(self, area):

        data = self.environment.areas[area]

        print("Vacuum is checking", area)

        if data["occupied"]:
            print(area, "is occupied. Skipping...")
            return

        if data["dirt"] > 0:

            print(area, "is Dirty")
            print("Dirt Level:", data["dirt"])

            data["dirt"] = 0

            self.battery -= 15
            self.score += 10

            print(area, "has been Cleaned")

        else:
            print(area, "is Already Clean")


    # Recharge vacuum cleaner battery
    def recharge(self):

        print("\nBattery is Low")

        print(
            "Vacuum is going to Laboratory Charging Station"
        )

        self.currentLocation = "Laboratory Charging Station"

        self.battery = 100

        print("Battery Charged to 100%")


    # Main laboratory cleaning process
    def clean_laboratory(self):

        print(
            "\nVacuum starts cleaning the laboratory"
        )

        while True:

            dirty_available_areas = [
                area
                for area in self.environment.areas
                if self.environment.areas[area]["dirt"] > 0
                and not self.environment.areas[area]["occupied"]
            ]

            if not dirty_available_areas:
                break

            if self.battery < 30:
                self.recharge()

            area = self.select_area()

            if area is None:
                break

            self.move_to(area)

            self.clean_area(area)

            print("Battery:", self.battery)

            print("Performance Score:", self.score)


    # Display final laboratory report
    def final_report(self):

        print("\nFinal Laboratory Environment")

        print("-----------------------------------")

        for area, data in self.environment.areas.items():

            print(
                area,
                "| Dirt:",
                data["dirt"],
                "| Occupied:",
                data["occupied"]
            )

        print(
            "\nFinal Location:",
            self.currentLocation
        )

        print(
            "Final Battery:",
            self.battery
        )

        print(
            "Final Performance Score:",
            self.score
        )


# Create Laboratory Environment
environment = LaboratoryEnvironment()


# Create Laboratory Vacuum Agent
agent = LaboratoryVacuumAgent(environment)


# Display initial environment
agent.display_environment()


# Start laboratory cleaning
agent.clean_laboratory()


# Display final report
agent.final_report()
```

Conclusion:
The Laboratory Vacuum Cleaner Agent was successfully implemented using Python and basic Artificial Intelligence concepts.

The system models a laboratory environment containing multiple areas with different dirt levels, priorities, and occupancy conditions. The vacuum cleaner observes the laboratory environment and intelligently selects a suitable area for cleaning instead of simply following a fixed sequence.

The agent avoids occupied laboratory areas, gives preference to areas with higher dirt levels and priorities, manages its battery by recharging when necessary, and maintains a performance score based on its actions.

Thus, this experiment demonstrates the basic working of an AI-based Simple Reflex Agent and provides a foundation for developing more advanced autonomous cleaning systems for laboratories, hospitals, industries, research centers, offices, and other large environments.
